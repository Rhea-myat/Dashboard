```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Job Field
df.jobfield.value_counts().plot(kind='bar', figsize=(10,6))
plt.xlabel('Job Field')
plt.ylabel('Count')
plt.title('Distribution of Job Fields')
plt.show()
```

```python
# Job Title
df.jobtitle.value_counts().head(20).plot(kind='bar', figsize=(10,6))
plt.xlabel('Job Title')
plt.ylabel('Count')
plt.title('Top 20 Job Titles')
plt.show()  
```

```python
# satisfaction
df.satisfied.value_counts().plot(kind='bar', figsize=(10,6))
plt.xlabel('Satisfaction Level')
plt.ylabel('Count')
plt.title('Distribution of Satisfaction Levels')
plt.show()
```

```python
# Psychotype vs Satisfaction
plt.figure(figsize=(12,6))
sns.countplot(x="psychotype", hue="satisfied", data=df)
plt.title("Satisfaction by Psychotype")
plt.xlabel("Psychotype")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.histplot(data=df, x='psychotype', hue='jobfield', multiple='stack', stat='count', shrink=0.8)
plt.xlabel('Psychotype')
plt.ylabel('Count')
plt.title('Distribution of Psychotypes by Job Field')
plt.xticks(rotation=45)
plt.show()
```

```python
# Drop rows with NaN in jobfield or psychotype before crosstab
filtered_df = df.dropna(subset=['jobfield', 'psychotype'])
pd.crosstab(filtered_df.jobfield, filtered_df.psychotype).plot(kind='bar', figsize=(10,6))
plt.title('Job Field vs Psychotype')
plt.xlabel('Job Field')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.legend(title='Psychotype')
plt.show()
```

```python
crosstab = pd.crosstab(df.jobfield, df.psychotype)
plt.figure(figsize=(10,6))
sns.heatmap(crosstab, annot=True, fmt='d', cmap='YlGnBu')
plt.title('Heatmap of Job Field vs Psychotype')
plt.xlabel('Psychotype')
plt.ylabel('Job Field')
plt.show()
```

```python
crosstab = pd.crosstab(df.jobfield, df.psychotype)
plt.figure(figsize=(10,6))
sns.heatmap(crosstab,  cmap='YlGnBu', annot=False)
plt.title('Heatmap of Job Field vs Psychotype')
plt.xlabel('Psychotype')
plt.ylabel('Job Field')
plt.show()
```

```python
df_pivot = df.pivot_table(index='jobfield', columns='psychotype', values='satisfied')
ax = sns.heatmap(df_pivot)
```

### Scales
1. I/E - Introversion/Extroversion
2. S/N - Sensing/Intuition 
3. T/F - Thinking/Feeling
4. J/P  - Judging/Perceiving 

```python
# scales 
import seaborn as sns

for col in [c for c in df.columns if c.startswith("scale_")]:
    plt.figure(figsize=(8, 4))
    sns.histplot(df[col], kde=True)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show()
```

### Relationship between two variables 

```python
pd.crosstab(df.jobfield, df.psychotype).plot(kind='bar', figsize=(10,6))
plt.title('Job Field vs Psychotype')
plt.xlabel('Job Field')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Psychotype', bbox_to_anchor=(1.05,1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
top_fields = df.jobfield.value_counts().head(10).index 
df_top = df[df.jobfield.isin(top_fields)]

plt.figure(figsize=(10,6))
sns.countplot(data=df_top, x='jobfield', hue='psychotype', palette='Set2')
plt.title('Psychotype Distribution in Top 10 Job Fields')
plt.xlabel('Job Field')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Psychotype', bbox_to_anchor=(1.05,1), loc='upper left')
plt.tight_layout()
plt.show()
```
